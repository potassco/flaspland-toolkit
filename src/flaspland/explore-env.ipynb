```python
# construct fact format for waypoints
# waypoint(Z,Ord,(X,Y),Dir,Arr,Dep).
dir_map = {0:"n", 1:"e", 2:"s", 3:"w"}

for agent_num, agent_info in enumerate(env.agents):
    num_wps = len(agent_info.waypoints)
    for wp in range(1,num_wps):
        Ord = wp
        Y,X = agent_info.waypoints[wp][0].position
        Dir = agent_info.waypoints[wp][0].direction
        Dep = agent_info.waypoints_earliest_departure[wp-1]
        Arr = agent_info.waypoints_latest_arrival[wp]
        print(f"waypoint({agent_num},{Ord},({Y},{X}),{dir_map[Dir]},{Arr},{Dep}). % waypoint(Z,Ord,(X,Y),Dir,Arr,Dep).")
        
            
```
